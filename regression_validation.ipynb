## **Model A:** `Hours_Studied` & `Exam_Score`

Testing **Ho: B1 = 0** (no linear relationship between weekly study hours and exam score).

```python
model_a = LinearRegressionStatisticalView(alpha=0.05)
model_a.fit(df['Hours_Studied'].values, df['Exam_Score'].values)
```

```python
f_a  = model_a.f_statistic()
fc_a = model_a.f_critical()

if f_a > fc_a:
    decision_a = 'Reject Ho'
else: 
    decision_a ='Fail to Reject Ho'

print(f'F-statistic : {f_a:.4f}')
print(f'F-critical  : {fc_a:.4f}')
print(f'Beta Confidence Intervals : {model_a.beta_confidence_intervals()}')
print(f'R-squared   : {model_a.r_squared():.4f}')
print(f'Pearson r   : {model_a.r():.4f}')
print(f'Decision    : {decision_a}')
```

```python
summary_a = model_a.summary()
summary_a
```

## **Model B:** `Attendance` & `Exam_Score`

Testing **Ho: B1 = 0** (no linear relationship between attendance rate and exam score).

```python
model_b = LinearRegressionStatisticalView(alpha=0.05)
model_b.fit(df['Attendance'].values, df['Exam_Score'].values)
```

```python
f_b  = model_b.f_statistic()
fc_b = model_b.f_critical()
if f_b > fc_b:
    decision_B = 'Reject Ho'
else: 
    decision_B = 'Fail to Reject Ho'

print(f'F-statistic : {f_b:.4f}')
print(f'F-critical  : {fc_b:.4f}')
print(f'Beta Confidence Intervals : {model_b.beta_confidence_intervals()}')
print(f'R-squared   : {model_b.r_squared():.4f}')
print(f'Pearson r   : {model_b.r():.4f}')
print(f'Decision    : {decision_B}')
```

```python
summary_b = model_b.summary()
summary_b
```

## **compare()** Function

```python
result = compare(
    df,
    x_col_1='Hours_Studied',
    x_col_2='Attendance',
    y_col='Exam_Score',
    alpha=0.05
)
result
```

### **Which predictor is superior based on the F-statistic?**
Based on F-statistic, `Attendance` is the superior predictor.

## **Verification**

```python
import statsmodels.api as sm

X_sm = sm.add_constant(df['Hours_Studied'].values)
ols_a = sm.OLS(df['Exam_Score'].values, X_sm).fit()

X_sm = sm.add_constant(df['Attendance'].values)
ols_b = sm.OLS(df['Exam_Score'].values, X_sm).fit()

sm_ci_a = ols_a.conf_int(alpha=0.05)
sm_ci_b = ols_b.conf_int(alpha=0.05)

for model_name, ols, sm_ci, m in [
    ("Model A: Hours_Studied", ols_a, sm_ci_a, model_a),
    ("Model B: Attendance", ols_b, sm_ci_b, model_b)
]:
    print(f"\n'{model_name}' Verification vs statsmodels OLS")
    print(f"{'Metric':<20} {'Custom':>15} {'Statsmodels':>15} {'Match':>7}")
    rows = [
        ("Bo",          m.beta_0,             ols.params[0]),
        ("B1",          m.beta_1,             ols.params[1]),
        ("SSE",         m.SSE,                ols.ssr),
        ("SSR",         m.SSR,                ols.ess),
        ("MSE",         m.MSE,                ols.mse_resid),
        ("F-statistic", m.f_statistic_value,  ols.fvalue),
        ("R-squared",   m.r_squared_value,    ols.rsquared),
        ("Bo CI lower", m.beta_0_lower_bound, sm_ci[0,0]),
        ("Bo CI upper", m.beta_0_upper_bound, sm_ci[0,1]),
        ("B1 CI lower", m.beta_1_lower_bound, sm_ci[1,0]),
        ("B1 CI upper", m.beta_1_upper_bound, sm_ci[1,1]),
    ]
    for metric, custom_val, sm_val in rows:
        match = "Passed" if abs(custom_val - sm_val) < 1e-6 else "Failed"
        print(f"{metric:<20} {custom_val:>15.6f} {sm_val:>15.6f} {match:>7}")
```
